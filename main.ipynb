```python
import numpy as np
import pandas as pd

data = pd.read_csv("C:\\Users\\vardh\\Desktop\\OPENCODEETASK\\fashion-mnist_train.csv")

```

```python


X = data.iloc[:, 1:].to_numpy(dtype=np.float32)  
y = data.iloc[:, 0].to_numpy()                   

X /= 255.0


X_train, y_train = X[:5000], y[:5000]
X_test, y_test = X[5000:6000], y[5000:6000]


def euclidean_distance(x1, x2):
    return np.sum((x1 - x2) ** 2)   


class KNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict_one(self, x):
        
        distances = np.sum((self.X_train - x) ** 2, axis=1)
        
        k_idx = np.argpartition(distances, self.k)[:self.k]
        k_labels = self.y_train[k_idx]
       
        values, counts = np.unique(k_labels, return_counts=True)
        return values[np.argmax(counts)]

    def predict(self, X):
        return np.array([self.predict_one(x) for x in X])


```

```python

knn = KNN(k=3)
knn.fit(X_train, y_train)

predictions = knn.predict(X_test)
accuracy = np.mean(predictions == y_test)
print("Accuracy:", accuracy*100,"%")
```
